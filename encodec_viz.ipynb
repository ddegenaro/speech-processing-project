```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
WRITE = True
```

```python
data = pd.DataFrame(
    json.load(open(os.path.join('encodec_eval', 'data.json'), 'r', encoding='utf-8'))
)
```

```python
data
```

```python
data['fname'] = data['from'].apply(lambda x: x.split('/')[-1])
```

```python
data['bitrate_encoded'] = data['encoder_rep_size'] * 8 / data['to_duration']
```

```python
fig = px.scatter(
    data,
    x = 'num_codebooks',
    y = 'audio_compression_ratio',
    color = 'fname',
    title = 'Audio compression ratio (↓) vs. # codebooks by audio file',
    width = 600,
    height = 400
)
fig.update_traces(marker=dict(size=8))
fig.update_layout(
    xaxis = dict(
        tickmode = 'array',
        tickvals = data['num_codebooks']
    )
)
if WRITE:
    fig.write_image(os.path.join('figs', 'audio_compression_ratio.pdf'), format='pdf')
fig.show()
```

```python
fig = px.scatter(
    data,
    x = 'num_codebooks',
    y = 'encoder_compression_ratio',
    color = 'fname',
    title = 'Encoder compression ratio (↓) vs. # codebooks by audio file',
    width = 600,
    height = 400
)
fig.update_traces(marker=dict(size=8))
fig.update_layout(
    xaxis = dict(
        tickmode = 'array',
        tickvals = data['num_codebooks']
    )
)
if WRITE:
    fig.write_image(os.path.join('figs', 'encoder_compression_ratio.pdf'), format='pdf')
fig.show()
```

```python
fig = px.scatter(
    data,
    x = 'num_codebooks',
    y = 'bitrate_kbps',
    color = 'fname',
    title = 'Bitrate (↓) vs. # codebooks by audio file',
    width = 600,
    height = 400
)
fig.update_traces(marker=dict(size=8))
fig.update_layout(
    xaxis = dict(
        tickmode = 'array',
        tickvals = data['num_codebooks']
    )
)
if WRITE:
    fig.write_image(os.path.join('figs', 'bitrate.pdf'), format='pdf')
fig.show()
```

```python
fig = px.scatter(
    data,
    x = 'num_codebooks',
    y = 'bitrate_encoded',
    color = 'fname',
    title = 'Bitrate of encoding (↓) vs. # codebooks by audio file',
    width = 600,
    height = 400
)
fig.update_traces(marker=dict(size=8))
fig.update_layout(
    xaxis = dict(
        tickmode = 'array',
        tickvals = data['num_codebooks']
    )
)
if WRITE:
    fig.write_image(os.path.join('figs', 'encoding_bitrate.pdf'), format='pdf')
fig.show()
```
